# ***Sample SuperStore Dataset Cleaning and Preprocessing***

The objective of this project is to analyze the Sample Superstore Dataset by performing data cleaning and data preprocessing,The project aims to understand the dataset, identify patterns, detect missing values and outliers and duplicates.

```python
import pandas as pd
import numpy as np
df=pd.read_csv("SampleSuperstore.csv")
print(df)
print(df.shape)
```

**Dataset Overview**

This section provides an overview of the dataset by displaying the first few rows, checking the dataset structure, identifying column names, and understanding the statistical summary.

```python
print("\n Data types and non null counts")
print(df.info)
print("statistical summary:")
print(df.describe(include="all").T)
```

**Data Preprocessing**

In this section, we check the dataset for missing values, duplicate records, and incorrect data types to ensure the data is clean and ready for analysis.

```python
print("\n missing values:")
print(df.isnull().sum())
print("\n exact duplicate rows:", df.duplicated().sum())
print("\n Duplicate row IDs:")
if "Row ID" in df.columns:
  print(df["Row Id"].duplicated().sum())


```

**Removing Duplicates**
Duplicate records can cause data to be counted more than once in analysis. Removing exact duplicates helps keep summaries and results accurate.


```python
df_clean = df.drop_duplicates()
print("\n cleaned dataset shape:", df_clean.shape)
```

After cleaning a dataset, it’s important to validate that the changes worked. This code checks for missing values and duplicate rows, then summarizes the most frequent values in each column. These checks help confirm data completeness and consistency and can reveal unexpected values that may need further review

```python
# Null values per column
print("Null values per column:")
print(df_clean.isnull().sum())

# Total null values
print("\nTotal null values:", df_clean.isnull().sum().sum())

# Exact duplicate rows
print("\nDuplicate rows:", df_clean.duplicated().sum())

# Unique values in each column — unusual values
for column in df_clean.columns:
    print(f"\n{column} — unique values:")
    print(df_clean[column].value_counts(dropna=False).head(20))
```

drop_duplicates() removes rows that are identical across all columns. The code then checks the cleaned DataFrame again: isnull().sum() reports missing values, duplicated().sum() confirms whether any exact duplicates remain, and len(df_clean) shows the number of rows left. If one duplicate row was present, the row count decreases by one.

```python
df_clean.info()
df_clean = df_clean.drop_duplicates().copy()
print("Nulls per column:\n", df_clean.isnull().sum())
print("Duplicate rows:", df_clean.duplicated().sum())
print("Rows remaining:", len(df_clean))
```

This code checks each text column for leading or trailing spaces and displays up to 10 example values. Extra spaces can make identical categories appear different, such as "West" and " West", which can affect grouping and analysis. The check also lets you review the text values for inconsistent capitalization; it does not change the data.

```python
# Check text values for leading/trailing spaces and inconsistent casing
for column in df_clean.select_dtypes(include="object").columns:
    values = df_clean[column].dropna().unique()
    print(f"\n{column}:")
    print("Examples:", values[:10])
    print("Has extra spaces:", any(value != value.strip() for value in values))
```

This code checks the numeric ranges and sample values of Sales, Profit, and Discount. The minimum and maximum help identify the range of values in each column, while the examples show how the data is stored. This helps confirm that Sales and Profit contain monetary amounts and that Discount is stored as a decimal rate, such as 0.2 for 20%

```python
for column in ["Sales", "Profit", "Discount"]:
    print(f"\n{column}")
    print("Minimum:", df_clean[column].min())
    print("Maximum:", df_clean[column].max())
    print("Examples:", df_clean[column].head().tolist())
```

The Discount column stores discount rates as decimals, such as 0.2 for 20%. This code multiplies those values by 100 and formats them with a % sign in a new Discount % column for easier reading. The original Discount values remain unchanged for calculations.

```python
# Create a display-only percentage column
df_clean["Discount %"] = (df_clean["Discount"] * 100).map(lambda value: f"{value:g}%")

print(df_clean[["Discount", "Discount %"]].head())
```

This code loads the original CSV and reads Postal Code as text so it stays an identifier. It removes exact duplicate rows and stores the result in df_clean

```python
df = pd.read_csv("SampleSuperstore.csv", dtype={"Postal Code": "string"})

df_clean = df.drop_duplicates().copy()
df_clean["Discount %"] = (df_clean["Discount"] * 100).map(
    lambda value: f"{value:g}%"
)
```

```python
output_file = "SampleSuperstore_data_set.csv"

df_clean.to_csv(output_file, index=False)

print(f"Cleaned dataset saved as: {output_file}")
```
